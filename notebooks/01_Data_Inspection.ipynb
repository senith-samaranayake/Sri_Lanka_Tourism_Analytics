```python

from pathlib import Path
import pandas as pd

# Project paths
PROJECT_DIR = Path.cwd().parent
RAW_DIR = PROJECT_DIR / "data" / "raw"
PROCESSED_DIR = PROJECT_DIR / "data" / "processed"

PROCESSED_DIR.mkdir(parents=True, exist_ok=True)

# Workbook configuration
files = {
    2022: {"file": "tourist_arrivals_2022.xlsx", "sheet": "Jan ~ Dec", "header": 2},
    2023: {"file": "tourist_arrivals_2023.xlsx", "sheet": "Sheet1", "header": 2},
    2024: {"file": "tourist_arrivals_2024.xlsx", "sheet": "Sheet1", "header": 1},
    2025: {"file": "tourist_arrivals_2025.xlsx", "sheet": "All Country", "header": 2},
    2026: {"file": "tourist_arrivals_2026.xlsx", "sheet": "All Country", "header": 2}
}

# Read and standardise each workbook
all_years = []

for year, config in files.items():

    file_path = RAW_DIR / config["file"]

    df = pd.read_excel(
        file_path,
        sheet_name=config["sheet"],
        header=config["header"]
    )

    # Remove completely empty columns
    df = df.dropna(axis=1, how="all")

    # Standardise column names
    df.columns = [str(col).strip() for col in df.columns]

    # Identify country and monthly columns
    country_col = next(
        col for col in df.columns
        if col.strip().lower() == "country"
    )

    month_aliases = {
        "jan": "January", "january": "January",
        "feb": "February", "february": "February",
        "mar": "March", "march": "March",
        "apr": "April", "april": "April",
        "may": "May",
        "jun": "June", "june": "June",
        "jul": "July", "july": "July",
        "aug": "August", "august": "August",
        "sep": "September", "sept": "September",
        "september": "September",
        "oct": "October", "october": "October",
        "nov": "November", "november": "November",
        "dec": "December", "december": "December"
    }

    month_columns = {
        col: month_aliases[col.strip().lower()]
        for col in df.columns
        if col.strip().lower() in month_aliases
    }

    # Keep only country and monthly arrival columns
    clean = df[[country_col] + list(month_columns.keys())].copy()
    clean = clean.rename(columns={country_col: "Country", **month_columns})

    # Remove empty country rows
    clean = clean.dropna(subset=["Country"])

    # Remove any accidental total/summary rows
    clean = clean[
        ~clean["Country"].astype(str).str.strip().str.lower().isin(
            ["total", "grand total"]
        )
    ]

    # Convert monthly values to numeric
    for month in month_columns.values():
        clean[month] = pd.to_numeric(clean[month], errors="coerce")

    # Convert wide format into monthly records
    long_df = clean.melt(
        id_vars=["Country"],
        var_name="Month",
        value_name="Arrivals"
    )

    long_df["Year"] = year
    long_df["Month_Number"] = pd.to_datetime(
        long_df["Month"], format="%B"
    ).dt.month

    long_df["Date"] = pd.to_datetime(
        dict(
            year=long_df["Year"],
            month=long_df["Month_Number"],
            day=1
        )
    )

    all_years.append(long_df)

# Combine all five years
tourism = pd.concat(all_years, ignore_index=True)

# Sort the data
tourism = tourism.sort_values(
    ["Year", "Month_Number", "Country"]
).reset_index(drop=True)

# Save integrated dataset
output_path = PROCESSED_DIR / "tourism_arrivals_monthly.csv"
tourism.to_csv(output_path, index=False)

print("Integration completed!")
print("Total records:", len(tourism))
print("Countries:", tourism["Country"].nunique())
print("Years:", sorted(tourism["Year"].unique()))
print("Date range:", tourism["Date"].min(), "to", tourism["Date"].max())

display(tourism.head(10))
```

```python

import pandas as pd
from pathlib import Path

# Load our integrated dataset
file_path = Path("../data/processed/tourism_arrivals_monthly.csv")

tourism = pd.read_csv(file_path, parse_dates=["Date"])

# 1. Dataset overview
print("DATASET OVERVIEW")
print("-" * 40)
print("Total records:", len(tourism))
print("Unique countries:", tourism["Country"].nunique())
print("Years:", sorted(tourism["Year"].unique()))

# 2. Records per year
print("\nRECORDS PER YEAR")
display(
    tourism.groupby("Year").agg(
        Records=("Country", "size"),
        Countries=("Country", "nunique"),
        Months=("Month_Number", "nunique"),
        Total_Arrivals=("Arrivals", "sum")
    )
)

# 3. Missing values
print("\nMISSING VALUES")
display(tourism.isnull().sum())

# 4. Duplicate records
print("\nDUPLICATE COUNTRY-MONTH RECORDS")
duplicates = tourism[
    tourism.duplicated(
        subset=["Country", "Year", "Month_Number"],
        keep=False
    )
]
display(duplicates)

# 5. Negative arrival values
print("\nNEGATIVE ARRIVAL VALUES")
display(tourism[tourism["Arrivals"] < 0])

# 6. Month coverage
print("\nMONTHS AVAILABLE PER YEAR")
display(
    tourism.groupby("Year")["Month_Number"]
    .apply(lambda x: sorted(x.unique()))
)

# 7. Country name variations
print("\nCOUNTRY NAMES")
display(
    tourism[["Country"]]
    .drop_duplicates()
    .sort_values("Country")
    .reset_index(drop=True)
)
```

```python
# Check missing arrivals
missing_arrivals = tourism[tourism["Arrivals"].isna()].copy()

print("Missing arrivals:", len(missing_arrivals))

display(
    missing_arrivals[
        ["Country", "Month", "Year", "Arrivals"]
    ].head(50)
)
```

```python
# Check suspicious rows where Country itself appears as a country name
country_rows = tourism[
    tourism["Country"].astype(str).str.strip().str.lower() == "country"
]

print("Suspicious 'Country' rows:", len(country_rows))

display(country_rows)
```

```python
# Check all country names
country_names = (
    tourism["Country"]
    .dropna()
    .astype(str)
    .str.strip()
    .drop_duplicates()
    .sort_values()
)

print("Number of unique country names:", len(country_names))

display(country_names.to_frame(name="Country"))
```

```python
# Count unwanted "Country" rows by year

country_header_rows = tourism[
    tourism["Country"].astype(str).str.strip().str.lower() == "country"
]

display(
    country_header_rows.groupby("Year").size().reset_index(name="Rows")
)
```

```python
# Check the genuine missing-arrival rows
# after excluding the unwanted "Country" rows

real_missing = tourism[
    tourism["Arrivals"].isna()
    & (tourism["Country"].astype(str).str.strip().str.lower() != "country")
]

print("Missing arrivals excluding 'Country' header rows:", len(real_missing))

display(real_missing)
```
